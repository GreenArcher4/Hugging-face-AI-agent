## Messages to prompt(tokenizing)


```python
messages = [
    {"role": "system", "content": "You are an AI assistant with access to various tools."},
    {"role": "user", "content": "Hi !"},
    {"role": "assistant", "content": "Hi human, what can help you with ?"},
]
```

* Convert conversation to a prompt: call fx apply_chat_template(), load it to tokenizer variable
* This apply_chat_template() function will be used in the backend of your API, when you interact with messages in the ChatML format.

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("HuggingFaceTB/SmolLM2-1.7B-Instruct")
rendered_prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
```

Huggingface's python library for working with a pre trainened LLM 

Autotokenizer is a helper class that autodetects which tokenizer a given model needs eg: LlamaTokenizerFast for Meta'z Ollama, GPT2TokenizerFast for Open AI's GPT 

.from_pretrained() downloads an already set up tokenizer for the model from its config files 

AutoTokenizer.from_pretrained("model  ID")- AutoTokenizer identifies the class/tokenizer, and .from_pretrained() is the action that fetches the details and builds the tokenizer. 

HuggingFaceTB/SmolLM2-1.7B-Instruct – hugging face model ID with the organization name, Small LLM family(SmolLm2),  with 1.7Billion parameters, and a fine tuned base model(instruct) 

apply_chat_template() a fx that applies a chat template 

tokenize=False –it returns a formatted prompt as a string. False is not the default(True) which would bring a list of integers 

This is helpful when you need to inspect the prompt or pass it to sth else that tokenizes 

add_generation_prompt=True - queues AI it is now its turn to start typing its response. Looks like: <|im_start|>assistant tart|>assistant tart|>assistant rt|>assistant stant ct) stant stant ct)
