3. Run Your First Tool Calling Agent
This agent uses WebSearchTool to search the web and write JSON blobs to solve the prompt:
Sample JSON BLOB:
Calling tool: 'web_search' with arguments: {'query': "best music recommendations for a party at Wayne's         │
│ mansion"}     

```python
from smolagents import ToolCallingAgent, WebSearchTool, LiteLLMModel

model=LiteLLMModel(
    model_id="ollama/qwen2.5-coder:1.5b",
    api_base="http://localhost:11434"
)

agent = ToolCallingAgent(tools=[WebSearchTool()],
                  model=model)

# Run the agent
agent.run("Search for the best music recommendations for a party at the Wayne's mansion.")
```
